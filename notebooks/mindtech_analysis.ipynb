```python
import pandas as pd
import numpy as np

# Prova prima il percorso standard o diretto
try:
    df = pd.read_csv('../data/Teen_Mental_Health_Dataset.csv')
except FileNotFoundError:
    try:
        df = pd.read_csv('data/Teen_Mental_Health_Dataset.csv')
    except FileNotFoundError:
        df = pd.read_csv('Teen_Mental_Health_Dataset.csv')

print("=== 1. STRUTTURA E ORGANIZZAZIONE DEL DATASET ===")
print(f"Dimensioni della matrice: {df.shape[0]} righe (soggetti), {df.shape[1]} colonne (variabili)")
print("\nTipi di dati e colonne:")
print(df.dtypes)

print("\n=== 2. VERIFICA DELL'INTEGRITÀ E CREDIBILITÀ DEI DATI ===")
missing_values = df.isnull().sum().sum()
duplicates = df.duplicated().sum()
print(f"Valori mancanti totali (Null): {missing_values}")
print(f"Record duplicati totali: {duplicates}")
print("Esito audit: Dataset perfettamente pulito, senza valori nulli o duplicati, garantendo un'elevata affidabilità della fonte.")

print("\n=== 3. STATISTICHE DESCRITTIVE PRELIMINARI ===")
print(df[['age', 'daily_social_media_hours', 'sleep_hours', 'stress_level']].describe())

print("\n=== 4. ORDINAMENTO E FILTRAGGIO AVANZATO ===")
heavy_users = df.sort_values(by='daily_social_media_hours', ascending=False).head(10)
print("\nTop 10 Heavy Users (maggior numero di ore sui social):")
print(heavy_users[['age', 'daily_social_media_hours', 'sleep_hours', 'stress_level', 'depression_label']])

at_risk_group = df[(df['sleep_hours'] <= 5) & (df['stress_level'] >= 7)]
print(f"\nNumero di soggetti nella fascia a rischio (sonno <= 5h e stress >= 7): {len(at_risk_group)}")
print("Media ore social nel gruppo a rischio:", round(at_risk_group['daily_social_media_hours'].mean(), 2))
print("Percentuale con indicatori di depressione nel gruppo a rischio:", round(at_risk_group['depression_label'].mean() * 100, 2), "%")
```

```python
import pandas as pd
import numpy as np

# 1. Caricamento sicuro del dataset (se non è già in memoria)
try:
    df = pd.read_excel('Teen_Mental_Health_Dataset.xlsx')
except FileNotFoundError:
    try:
        df = pd.read_csv('../data/Teen_Mental_Health_Dataset.csv')
    except FileNotFoundError:
        try:
            df = pd.read_csv('data/Teen_Mental_Health_Dataset.csv')
        except FileNotFoundError:
            df = pd.read_csv('Teen_Mental_Health_Dataset.csv')

# 2. Aggregazione per piattaforma social (Task 1)
platform_aggregation = df.groupby('platform_usage')[
    ['daily_social_media_hours', 'sleep_hours', 'stress_level', 'anxiety_level', 'addiction_level']
].mean().round(2)

print("=== AGGREGAZIONE PER PIATTAFORMA SOCIAL ===")
print(platform_aggregation)

# 3. Aggregazione per livello di interazione sociale
social_aggregation = df.groupby('social_interaction_level')[
    ['stress_level', 'anxiety_level', 'academic_performance']
].mean().round(2)

print("\n=== AGGREGAZIONE PER LIVELLO DI INTERAZIONE SOCIALE ===")
print(social_aggregation)
```

```python
# ==========================================
# TASK 2: ORGANIZZAZIONE E FORMATTAZIONE DEI DATI
# ==========================================

import pandas as pd

# 1. Creazione di una nuova categoria basata sulle ore di sonno
df['sleep_category'] = df['sleep_hours'].apply(
    lambda x: 'Insufficiente (<=5h)' if x <= 5.0 else 'Adeguata (>5h)'
)

# 2. Creazione di un indicatore binario di rischio (Stress e Ansia elevati >= 7)
df['high_stress_anxiety'] = (
    (df['stress_level'] >= 7) & (df['anxiety_level'] >= 7)
).astype(int)

# 3. Ordinamento strutturato del dataset per livello di dipendenza e ore social
df_formatted = df.sort_values(
    by=['addiction_level', 'daily_social_media_hours'], 
    ascending=[False, False]
)

print("=== ANTEPRIMA DATI RIFORMATTATI E NUOVE VARIABILI ===")
print(df_formatted[['age', 'daily_social_media_hours', 'sleep_category', 'stress_level', 'anxiety_level', 'high_stress_anxiety']].head(10))

# Verifichiamo la distribuzione della nuova categoria di sonno
print("\n=== DISTRIBUZIONE CATEGORIA SONNO ===")
print(df['sleep_category'].value_counts())

# Verifichiamo quanti soggetti rientrano nel flag di rischio combinato
print("\n=== SOGGETTI CON ALTO STRESS E ANSIA (0 = No, 1 = Sì) ===")
print(df['high_stress_anxiety'].value_counts())
```

```python
# ==========================================
# TASK 3: ESECUZIONE DEI CALCOLI STATISTICI E CORRELAZIONI
# ==========================================

import pandas as pd

print("=== 1. MATRICE DI CORRELAZIONE TRA VARIABILI CHIAVE ===")
# Selezioniamo le variabili numeriche rilevanti per l'indagine psicologica e digitale
numeric_vars = [
    'daily_social_media_hours', 'sleep_hours', 'screen_time_before_sleep', 
    'stress_level', 'anxiety_level', 'addiction_level', 'academic_performance'
]

# Calcoliamo la matrice di correlazione arrotondata a 2 decimali
correlation_matrix = df[numeric_vars].corr().round(2)
print(correlation_matrix)

print("\n=== 2. IMPATTO DELLO SCHERMO PRIMA DI DORMIRE SUL SONNO E SULL'ANSIA ===")
# Calcoliamo le medie raggruppando per fasce di tempo trascorso davanti allo schermo prima di dormire
screen_impact = df.groupby('screen_time_before_sleep')[
    ['sleep_hours', 'anxiety_level', 'stress_level']
].mean().round(2)

print(screen_impact)
```
