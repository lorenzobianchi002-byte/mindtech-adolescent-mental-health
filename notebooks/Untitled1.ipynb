```python
import pandas as pd
import numpy as np

# 1. Caricamento del dataset dalla cartella corrente
df = pd.read_csv('Teen_Mental_Health_Dataset.csv')

def perform_preliminary_validation(dataframe):
    """
    Esegue l'ispezione strutturale e la validazione preliminare 
    del dataset per verificarne l'integrità e la qualità dei dati.
    """
    print("==================================================")
    print("      VALUTAZIONE PRELIMINARE DEL DATASET         ")
    print("==================================================")
    
    # Controllo delle dimensioni (soggetti e variabili)
    rows, cols = dataframe.shape
    print(f"\n[1. Dimensioni del Campione]")
    print(f" - Numero di osservazioni (soggetti): {rows}")
    print(f" - Numero di variabili (colonne): {cols}")
    
    # Tassonomia e tipi di dato
    print(f"\n[2. Tassonomia e Tipi di Dato]")
    display(dataframe.dtypes)
    
    # Verifica dei valori nulli e mancanti
    null_counts = dataframe.isnull().sum()
    total_nulls = null_counts.sum()  # <-- Corretto qui
    print(f"\n[3. Controllo Integrità - Valori Mancanti]")
    if total_nulls > 0:
        print(f" - Attenzione: Rilevati {total_nulls} valori nulli totali.")
        print(null_counts[null_counts > 0])
    else:
        print(" - Esito positivo: Nessun valore nullo riscontrato nel dataset.")
        
    # Controllo di record duplicati
    duplicates = dataframe.duplicated().sum()
    print(f"\n[4. Controllo Unicità]")
    print(f" - Record duplicati individuati: {duplicates}")
    
    # Statistiche descrittive preliminari per le variabili numeriche
    print(f"\n[5. Statistiche Descrittive Preliminari]")
    display(dataframe.describe())
    print("==================================================")

# Esecuzione della validazione preliminare
perform_preliminary_validation(df)
```
